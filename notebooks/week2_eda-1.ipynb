# Week 2 – Exploratory Data Analysis and Visualization

## Laptop Price Dataset

This notebook performs EDA on the cleaned Week 1 laptop dataset using **Pandas, NumPy, Matplotlib, and Seaborn**.

### Analysis covered
- Dataset structure and quality
- Descriptive statistics
- Brand, laptop type, OS and CPU distributions
- Price distribution
- RAM vs Price
- Screen Size vs Price
- Price by Laptop Type
- Average Price by Brand
- Correlation analysis
- Touchscreen and IPS analysis


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)

# GitHub repository path:
df = pd.read_csv("../data/cleaned/laptop_cleaned_week1.csv")

print("Dataset loaded successfully")
print("Shape:", df.shape)
```

## 1. Initial Data Exploration

```python
df.head()
```

```python
df.info()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nColumns:")
print(df.columns.tolist())
```

```python
print("Missing values:")
display(df.isnull().sum().sort_values(ascending=False))

print("Duplicate rows:", df.duplicated().sum())
```

## 2. Descriptive Statistics

```python
df.describe().T
```

## 3. Categorical Feature Analysis

```python
print("Top brands:")
display(df["Company"].value_counts().head(15))

print("Laptop types:")
display(df["TypeName"].value_counts())

print("Operating systems:")
display(df["OpSys"].value_counts())

print("CPU brands:")
display(df["CPU_Brand"].value_counts())
```

## 4. Brand Distribution

```python
brand_counts = df["Company"].value_counts()

plt.figure(figsize=(9,5))
brand_counts.head(10).sort_values().plot(kind="barh")
plt.title("Top 10 Laptop Brands by Number of Records")
plt.xlabel("Number of Laptops")
plt.ylabel("Brand")
plt.tight_layout()
plt.show()
```

### Interpretation
The chart shows which manufacturers are most represented in this dataset. The distribution is not uniform, so it should not be interpreted as real-world market share.


## 5. Price Distribution

```python
plt.figure(figsize=(9,5))
sns.histplot(df["Price"].dropna(), bins=30, kde=True)
plt.title("Laptop Price Distribution")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

print("Mean price:", round(df["Price"].mean(), 2))
print("Median price:", round(df["Price"].median(), 2))
```

### Interpretation
The histogram shows the spread and shape of laptop prices. Comparing mean and median helps identify skewness and the influence of expensive laptops.


## 6. Laptop Type Distribution

```python
plt.figure(figsize=(9,5))
sns.countplot(data=df, x="TypeName", order=df["TypeName"].value_counts().index)
plt.title("Laptop Type Distribution")
plt.xlabel("Laptop Type")
plt.ylabel("Number of Laptops")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

### Interpretation
Different laptop categories have different representation. Small categories should be interpreted carefully because their summary statistics may be less stable.


## 7. RAM vs Price

```python
ram_corr = df[["Ram_GB", "Price"]].corr().iloc[0,1]

plt.figure(figsize=(9,5))
sns.scatterplot(data=df, x="Ram_GB", y="Price", alpha=0.5)
plt.title(f"RAM vs Laptop Price (Correlation = {ram_corr:.2f})")
plt.xlabel("RAM (GB)")
plt.ylabel("Price")
plt.tight_layout()
plt.show()

print("RAM-Price correlation:", round(ram_corr, 3))
```

### Interpretation
A positive correlation indicates that higher-RAM laptops tend to have higher prices in this sample. RAM is not the only factor: CPU, GPU, storage, display, brand and laptop type can also affect price.


## 8. Screen Size vs Price

```python
screen_corr = df[["Inches", "Price"]].corr().iloc[0,1]

plt.figure(figsize=(9,5))
sns.scatterplot(data=df, x="Inches", y="Price", alpha=0.5)
plt.title(f"Screen Size vs Laptop Price (Correlation = {screen_corr:.2f})")
plt.xlabel("Screen Size (inches)")
plt.ylabel("Price")
plt.tight_layout()
plt.show()

print("Screen-Price correlation:", round(screen_corr, 3))
```

### Interpretation
Screen size has a weaker relationship with price than RAM. Laptops with similar screen sizes can have very different prices because their other specifications differ.


## 9. Price by Laptop Type

```python
major_types = df["TypeName"].value_counts().head(7).index

plt.figure(figsize=(10,5))
sns.boxplot(data=df[df["TypeName"].isin(major_types)],
            x="TypeName", y="Price")
plt.title("Laptop Price by Major Laptop Type")
plt.xlabel("Laptop Type")
plt.ylabel("Price")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

### Interpretation
The box plot compares price ranges across categories. Gaming, workstation and premium-oriented products can have higher prices because of stronger hardware and specialized features.


## 10. Average Price by Brand

```python
avg_price_brand = (
    df.groupby("Company")["Price"]
      .agg(["count", "mean", "median"])
      .sort_values("mean", ascending=False)
)

display(avg_price_brand.head(10))
```

```python
top_avg_brands = avg_price_brand.head(10).sort_values("mean")

plt.figure(figsize=(9,5))
top_avg_brands["mean"].plot(kind="barh")
plt.title("Top 10 Brands by Average Laptop Price")
plt.xlabel("Average Price")
plt.ylabel("Brand")
plt.tight_layout()
plt.show()
```

### Interpretation
Average price differs between manufacturers. This does not prove a brand causes higher prices because each brand has a different product mix.


## 11. Operating System Distribution

```python
plt.figure(figsize=(9,5))
sns.countplot(data=df, x="OpSys", order=df["OpSys"].value_counts().index)
plt.title("Operating System Distribution")
plt.xlabel("Operating System")
plt.ylabel("Number of Laptops")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

### Interpretation
The chart identifies the dominant operating systems in the dataset. Rare operating-system categories have fewer observations and should be compared cautiously.


## 12. CPU Brand Distribution

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="CPU_Brand", order=df["CPU_Brand"].value_counts().index)
plt.title("CPU Brand Distribution")
plt.xlabel("CPU Brand")
plt.ylabel("Number of Laptops")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

### Interpretation
The CPU distribution shows which processor brands dominate the sample. The distribution is useful context when explaining differences in laptop prices.


## 13. Correlation Matrix

```python
numeric_cols = [
    c for c in [
        "Price", "Inches", "Ram_GB", "Weight_kg",
        "CPU_GHz", "ScreenWidth_px", "ScreenHeight_px"
    ] if c in df.columns
]

corr = df[numeric_cols].corr()

plt.figure(figsize=(9,7))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation Matrix of Numerical Features")
plt.tight_layout()
plt.show()
```

### Interpretation
The correlation matrix summarizes linear relationships among numerical features. Correlation helps identify candidate relationships but does not prove causation.


## 14. Touchscreen Analysis

```python
display(df.groupby("Touchscreen")["Price"].agg(["count","mean","median"]))

plt.figure(figsize=(7,5))
sns.barplot(data=df, x="Touchscreen", y="Price", estimator="mean")
plt.title("Average Laptop Price by Touchscreen Availability")
plt.xlabel("Touchscreen (0 = No, 1 = Yes)")
plt.ylabel("Average Price")
plt.tight_layout()
plt.show()
```

### Interpretation
Touchscreen laptops may have different average prices because touchscreen functionality is often combined with convertible designs and premium displays. The chart does not establish a causal effect.


## 15. IPS Display Analysis

```python
display(df.groupby("IPS")["Price"].agg(["count","mean","median"]))

plt.figure(figsize=(7,5))
sns.barplot(data=df, x="IPS", y="Price", estimator="mean")
plt.title("Average Laptop Price by IPS Display")
plt.xlabel("IPS (0 = No, 1 = Yes)")
plt.ylabel("Average Price")
plt.tight_layout()
plt.show()
```

### Interpretation
IPS-equipped laptops can have different average prices because display technology may be associated with premium product positioning. Other specifications also contribute to price.


## 16. Key Findings

```python
print("EDA SUMMARY")
print("=" * 50)
print("Records:", len(df))
print("Variables:", df.shape[1])
print("Mean price:", round(df["Price"].mean(), 2))
print("Median price:", round(df["Price"].median(), 2))
print("Most common brand:", df["Company"].mode().iloc[0])
print("Most common laptop type:", df["TypeName"].mode().iloc[0])
print("Most common OS:", df["OpSys"].mode().iloc[0])
print("RAM-Price correlation:", round(df[["Ram_GB","Price"]].corr().iloc[0,1], 3))
print("Screen-Price correlation:", round(df[["Inches","Price"]].corr().iloc[0,1], 3))
```

## 17. Conclusion

The EDA shows that laptop price is a multi-factor outcome. RAM has a positive relationship with price, while screen size has a weaker relationship. Laptop type, brand, operating system, CPU brand, touchscreen availability and IPS display characteristics provide additional context.

The visualizations provide evidence for the numerical findings and make trends and distributions easier to interpret. These findings are exploratory and should not be treated as causal conclusions. A future stage could use regression or machine-learning models to predict laptop prices from multiple specifications.
